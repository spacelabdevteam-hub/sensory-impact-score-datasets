```python
from datasets import load_dataset
import pandas as pd

# Explicitly target the flat CSV file so Hugging Face `datasets` never collides with nested JSON metadata
ds = load_dataset('spacejourney/rome-sensory-impact-dataset', data_files='data/rome_sensory_impact_sis_v1.csv')
df = pd.DataFrame(ds['train'])

print(f'✅ Successfully loaded {len(df)} contemplative sanctuaries from spacejourney/rome-sensory-impact-dataset')
cols = [c for c in ['name', 'category', 'neighborhood', 'sis_score', 'acoustic_noise_db', 'golden_hour_window'] if c in df.columns]
df[cols].head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6), dpi=120)
sns.set_theme(style='darkgrid')

# Verified Solarpunk palette matching exact CSV category values
palette_map = {'Curated Premium': '#10B981', 'Free Restorative': '#E8C57D', 'Secret Sanctuary': '#34D399'}
scatter = sns.scatterplot(
    data=df,
    x='acoustic_noise_db',
    y='sis_score',
    hue='category',
    palette=palette_map,
    s=150,
    edgecolor='black',
    linewidth=1.2
)
plt.title('Rome, Italy: Acoustic Pressure (dB) vs. Sensory Impact Score (SIS)', fontsize=13, weight='bold', pad=14)
plt.xlabel('Acoustic Level (dB — Lower = Quieter Sanctuary)', fontsize=11)
plt.ylabel('Sensory Impact Score (SIS 0-100)', fontsize=11)
plt.axvline(x=53, color='#059669', linestyle='--', linewidth=1.5, label='WHO Restorative Guideline (53 dB Lden)')

for _, row in df.sort_values('sis_score', ascending=False).head(6).iterrows():
    short_name = str(row['name']).split(':')[0].split('|')[0][:26]
    plt.annotate(
        short_name, (row['acoustic_noise_db'], row['sis_score']),
        textcoords='offset points', xytext=(0, 9), ha='center', fontsize=8, weight='semibold'
    )

plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
top_5 = df.sort_values(by='sis_score', ascending=False).head(5)

print('🏆 TOP 5 RESTORATIVE SANCTUARIES — ROME, ITALY')
print('=' * 75)
for idx, (_, row) in enumerate(top_5.iterrows(), 1):
    print(f"\n{idx}. {row['name']} (SIS: {row['sis_score']}/100 | {row['acoustic_noise_db']} dB)")
    print(f"   Neighborhood: {row.get('neighborhood', 'Rome, Italy')} | Optimal Window: {row.get('golden_hour_window', 'Golden Hour')}")
    print(f"   🔗 Official Sanctuary Guide: {row.get('canonical_url', 'https://spacejourney.app')}")

print('\n✨ 1-Click Auto-Import Itinerary into Space Journey Planner:')
print('https://spacejourney.app/planificador-de-viajes/?ai_city=roma&ai_days=3&utm_source=colab_notebook')
```
